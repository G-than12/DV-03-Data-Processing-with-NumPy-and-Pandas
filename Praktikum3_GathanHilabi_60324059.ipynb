### 1. Menyiapkan Notebook dan Mengimpor Library

```python
import numpy as np
import pandas as pd
```

### 2. Dasar Array NumPy

```python
# Membuat array NumPy dari sebuah list
nilai_ujian = np.array([78, 85, 90, 65, 72])

print("Rata-rata:", np.mean(nilai_ujian))
print("Maksimum:", np.max(nilai_ujian))
print("Minimum:", np.min(nilai_ujian))
print("Std Deviasi:", np.std(nilai_ujian))
```

### 3. Membaca Dataset dengan Pandas

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
df.head() # menampilkan 5 baris pertama
```

### 4. Eksplorasi Awal Dataset

```python
print(df.shape) # jumlah baris & kolom
print(df.columns) # nama-nama kolom

df.info() # tipe data & jumlah data tidak kosong
df.describe() # statistik deskriptif kolom numerik
```

### 5. Mengakses Kolom dan Baris

```python
# Mengakses satu kolom
df["age"].head()

# Mengakses beberapa kolom sekaligus
df[["age", "sex", "survived"]].head()

# Memfilter baris berdasarkan kondisi
penumpang_anak = df[df["age"] < 18]
print("Jumlah penumpang anak:", len(penumpang_anak))
```

### 6. Mendeteksi Missing Value

```python
print(df.isnull().sum())
```

### 7. Menangani Missing Value

```python
# Strategi 1: isi nilai kosong dengan rata-rata (untuk data numerik)
df["age"] = df["age"].fillna(df["age"].mean())

# Strategi 2: hapus baris yang datanya kosong pada kolom tertentu
df = df.dropna(subset=["embarked"])

print(df.isnull().sum()) # periksa kembali
```

### 8. Mendeteksi dan Menghapus Duplikasi

```python
print("Jumlah data sebelum:", len(df))
print("Jumlah duplikasi:", df.duplicated().sum())

df = df.drop_duplicates()
print("Jumlah data sesudah:", len(df))
```

### 9. Menyimpan Dataset yang Sudah Dibersihkan

```python
df.to_csv("titanic_bersih.csv", index=False)
```

**Latihan 2:** Hitung rata-rata usia penumpang menggunakan NumPy (np.mean(df['age'])) dan Pandas (df['age'].mean()). Bandingkan apakah hasilnya sama.

```python
mean_np = np.mean(df['age'])
mean_pd = df['age'].mean()
print("Rata-rata (NumPy):", mean_np)
print("Rata-rata (Pandas):", mean_pd)
print("Apakah sama?", mean_np == mean_pd)
```

**Latihan 3:** Tampilkan data penumpang yang berada di kelas 1 (pclass == 1) dan selamat (survived == 1) menggunakan filter kondisi ganda (gunakan operator &).

```python
penumpang_selamat_kelas1 = df[(df['pclass'] == 1) & (df['survived'] == 1)]
display(penumpang_selamat_kelas1.head())
```

**Latihan 4:** Isi missing value pada kolom embarked dengan nilai yang paling sering muncul (modus), menggunakan df['embarked'].mode()[0].

```python
# Catatan: karena baris dengan embarked kosong sudah didrop sebelumnya, baris ini sebagai latihan logika saja.
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
print(df.isnull().sum())
```

**Latihan 5:** Hitung berapa persentase penumpang yang selamat dari keseluruhan data, menggunakan df['survived'].mean() * 100.

```python
persentase_selamat = df['survived'].mean() * 100
print(f"Persentase penumpang selamat: {persentase_selamat:.2f}%")
```

## H. Pertanyaan Diskusi

**1. Mengapa penting memeriksa dan menangani missing value sebelum data digunakan untuk membuat visualisasi?**
Karena data yang hilang dapat menyebabkan kesalahan dalam analisis atau visualisasi, mengganggu perhitungan statistik, atau menyebabkan fungsi/library plotting error. Menangani missing value memastikan visualisasi merepresentasikan data dengan akurat tanpa adanya bias atau kegagalan komputasi.

**2. Dalam kondisi seperti apa sebaiknya kita menggunakan dropna(), dan dalam kondisi seperti apa lebih baik menggunakan fillna()?**
- Gunakan `dropna()` saat jumlah missing value sangat sedikit dibandingkan total data, atau jika nilai tersebut krusial (misal label target) sehingga menebak nilainya berisiko menghasilkan kesimpulan yang salah.
- Gunakan `fillna()` saat kita masih ingin mempertahankan data tersebut (karena fiturnya penting atau jumlah observasi sedikit), dan kita bisa menggunakan pendekatan statistik (seperti rata-rata, median, modus) untuk mengisi kekosongan tanpa banyak mengubah distribusi data.

**3. Apa risiko yang bisa terjadi jika data duplikat tidak dihapus sebelum data dianalisis atau divisualisasikan?**
Data duplikat dapat memberikan bobot berlebih pada observasi tertentu, mendistorsi hasil analisis, statistik deskriptif, dan juga bentuk visualisasi (misalnya histogram akan memiliki frekuensi yang berlebihan di titik tertentu), sehingga kesimpulan yang ditarik menjadi bias dan kurang akurat.
