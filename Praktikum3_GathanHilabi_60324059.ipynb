### 1. Menyiapkan Notebook dan Mengimpor Library
Mengimpor pustaka `numpy` dan `pandas` untuk analisis dan manipulasi data tabular.


```python
import numpy as np
import pandas as pd
```

### 2. Dasar Array NumPy
Membuat array numerik dan menghitung ringkasan statistik dasar (rata-rata, maksimum, minimum, deviasi standar).


```python
# Membuat array NumPy dari sebuah list
nilai_ujian = np.array([78, 85, 90, 65, 72])

print("Rata-rata:", np.mean(nilai_ujian))
print("Maksimum:", np.max(nilai_ujian))
print("Minimum:", np.min(nilai_ujian))
print(f"Std Deviasi: {np.std(nilai_ujian):.2f}")
```

### 3. Membaca Dataset dengan Pandas
Memuat Titanic dataset langsung dari URL publik GitHub (seaborn-data) dan menampilkan 5 baris pertama.


```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
df.head() # menampilkan 5 baris pertama
```

### 4. Eksplorasi Awal Dataset
Mengetahui dimensi dataset, daftar kolom, tipe data masing-masing kolom, dan statistik deskriptif numerik.


```python
print(df.shape) # jumlah baris & kolom
print(df.columns) # nama-nama kolom
df.info() # tipe data & jumlah data tidak kosong
df.describe() # statistik deskriptif kolom numerik
```

### 5. Mengakses Kolom dan Baris
Mengambil kolom spesifik, memilih beberapa atribut sekaligus, dan melakukan filtering baris dengan boolean indexing.


```python
# Mengakses satu kolom
print("Contoh kolom age:")
print(df["age"].head())

# Mengakses beberapa kolom sekaligus
print("\nContoh subset kolom age, sex, survived:")
print(df[["age", "sex", "survived"]].head())

# Memfilter baris berdasarkan kondisi (usia < 18)
penumpang_anak = df[df["age"] < 18]
print("\nJumlah penumpang anak:", len(penumpang_anak))
```

### 6. Mendeteksi Missing Value
Mengecek jumlah nilai kosong (null / NaN) pada masing-masing kolom dataset.


```python
print("Jumlah missing value per kolom:")
print(df.isnull().sum())
```

### 7. Menangani Missing Value
Menerapkan dua strategi pembersihan missing value: imputasi rata-rata pada kolom `age` dan penghapusan baris pada kolom `embarked`.


```python
# Strategi 1: isi nilai kosong dengan rata-rata (untuk data numerik age)
df["age"] = df["age"].fillna(df["age"].mean())

# Strategi 2: hapus baris yang datanya kosong pada kolom tertentu (embarked)
df = df.dropna(subset=["embarked"])

print("Jumlah missing value setelah penanganan:")
print(df.isnull().sum()) # periksa kembali
```

### 8. Mendeteksi dan Menghapus Duplikasi
Mendeteksi apakah terdapat baris duplikat, lalu menghapusnya dengan `drop_duplicates()`.


```python
print("Jumlah data sebelum:", len(df))
print("Jumlah duplikasi:", df.duplicated().sum())

df = df.drop_duplicates()
print("Jumlah data sesudah:", len(df))
```

### 9. Menyimpan Dataset yang Sudah Dibersihkan
Menyimpan DataFrame hasil pembersihan ke dalam file `titanic_bersih.csv`.


```python
df.to_csv("titanic_bersih.csv", index=False)
print("Dataset bersih berhasil disimpan ke titanic_bersih.csv dengan total baris:", len(df))
```

Isi missing value pada kolom `embarked` dengan nilai yang paling sering muncul (modus), menggunakan `df['embarked'].mode()[0]`.


```python
# Menunjukkan teknik pengisian modus pada kolom kategorikal (embarked)
# Menggunakan dataset contoh untuk mendemonstrasikan penanganan missing value secara eksplisit
df_latihan = pd.read_csv(url)
modus_embarked = df_latihan["embarked"].mode()[0]
print(f"Modus kolom embarked: '{modus_embarked}'")
print("Missing value sebelum imputasi:", df_latihan["embarked"].isnull().sum())

# Mengisi missing value dengan nilai modus
df_latihan["embarked"] = df_latihan["embarked"].fillna(modus_embarked)
print("Missing value setelah imputasi:", df_latihan["embarked"].isnull().sum())

# Terapkan juga pada dataframe utama jika masih ada nilai kosong
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
```

## H. Pertanyaan Diskusi

### 1. Mengapa penting memeriksa dan menangani missing value sebelum data digunakan untuk membuat visualisasi?
**Jawaban:**
Memeriksa dan menangani *missing value* sangat penting karena:
1. **Mencegah Error dan Inkonsistensi Visualisasi:** Banyak pustaka visualisasi (seperti Matplotlib, Seaborn, Plotly) atau model machine learning dapat mengalami *error*, melewatkan baris tertentu secara diam-diam, atau menghasilkan agregasi statistik yang terdistorsi saat menemui nilai `NaN`.
2. **Menjaga Akurasi dan Validitas Representasi Data:** Data yang hilang secara sistematis (misalnya kelompok demografi tertentu tidak mengisi form) dapat menimbulkan bias interpretasi jika tidak ditangani dengan sengaja.
3. **Transparansi Integritas Data:** Mengetahui persentase missing value membantu analis memutuskan apakah variabel tersebut masih layak dipertahankan atau sebaiknya dieliminasi.

### 2. Dalam kondisi seperti apa sebaiknya kita menggunakan `dropna()`, dan dalam kondisi seperti apa lebih baik menggunakan `fillna()`?
**Jawaban:**
- **Gunakan `dropna()` ketika:**
  - Proporsi baris yang memiliki nilai kosong sangat kecil (misalnya < 1-2% dari total dataset), sehingga penghapusannya tidak akan mengubah distribusi data secara signifikan (contoh: 2 baris pada kolom `embarked`).
  - Kolom tersebut merupakan variabel target/kunci yang tidak boleh diasumsikan atau ditebak.
  - Kolom memiliki persentase missing value yang terlampau tinggi (misalnya > 70-80% seperti kolom `deck` pada Titanic), sehingga kolom tersebut lebih baik di-drop sepenuhnya.
- **Gunakan `fillna()` ketika:**
  - Ukuran dataset relatif terbatas sehingga membuang baris akan mengurangi ukuran sampel secara drastis.
  - Missing value dapat diestimasi secara logis menggunakan parameter statistik tanpa merusak distribusi data, seperti menggunakan **mean** atau **median** untuk fitur numerik yang mendekati normal/miring, atau **modus** untuk fitur kategorikal.

### 3. Apa risiko yang bisa terjadi jika data duplikat tidak dihapus sebelum data dianalisis atau divisualisasikan?
**Jawaban:**
Jika data duplikat tidak dihapus, dampak negatif yang dapat terjadi meliputi:
1. **Distorsi Statistik Deskriptif:** Nilai rata-rata, deviasi standar, kuartil, dan total agregasi akan terboboti secara berlebihan ke observasi yang terduplikasi (*pseudoreplication*).
2. **Bias Visualisasi:** Plot seperti histogram, boxplot, heatmap korelasi, dan countplot akan menampilkan densitas atau frekuensi palsu yang tidak mencerminkan populasi sebenarnya.
3. **Overfitting pada Pemodelan:** Jika data duplikat terbawa ke pemodelan prediktif, model dapat menghafal data duplikat tersebut sehingga terjadi kebocoran data (*data leakage*) antara data latih dan data uji.
